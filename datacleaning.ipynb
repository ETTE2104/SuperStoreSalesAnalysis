SUPERSTORE SALES DATA ANALYST

```python
# Setup thư viện
import pandas as pd 
```

```python
# Load file dữ liệu
df = pd.read_csv(r"C:\Users\ADMIN\Desktop\Đồ Án Kì\Datasets\Superstore Dataset.csv")
# Hiển thị toàn bộ dataframe
df
```

```python
#Xem thông tin dữ liệu
df.info() 
```

```python
#Đổi lại kiểu dữ liệu cho cột Order Date và Ship Date
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
#Xem lại thông tin dữ liệu sau khi đổi kiểu
df.info()
```

```python
df.describe()
```

```python
#Sắp xếp dữ liệu theo cột Order Date
df=df.sort_values(by="Order Date")

#Kiểm tra logic của dữ liệu
#Loại bỏ các dòng có ngày giao hàng trước ngày đặt hàng
df = df[df['Ship Date'] >= df['Order Date']]
df
```

```python
#Kiểm tra dữ liệu bị thiếu
df.isnull().sum() 
```

Dataset không có giá trị bị thiếu trong bất kỳ cột nào. Không cần xử lý thiếu dữ liệu.


```python
#Kiểm tra giá trị trùng lặp
df.duplicated().sum() 
```

```python
#Hiển thị các dòng bị trùng lặp
df[df.duplicated(keep=False)] 
```
